```python
import sqlite3
import pandas as pd

db_path = 'chembl_36/chembl_36_sqlite/chembl_36.db' 
conn = sqlite3.connect(db_path)

print("--- 질병 이름(Heading)과 MeSH ID 매칭 확인 ---\n")

query = """
SELECT DISTINCT
    mesh_id AS mesh_id,
    mesh_heading AS mesh_heading,
    efo_id AS efo_id,
    efo_term AS efo_term
FROM
    drug_indication
WHERE
    mesh_id IS NOT NULL
"""


df_mesh = pd.read_sql_query(query, conn)
conn.close()

print(df_mesh)
```

```python
df_mesh["mesh_heading"] = df_mesh['mesh_heading'].str.replace('"', '', regex=False)
```
