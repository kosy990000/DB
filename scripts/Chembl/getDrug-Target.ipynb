```python
import sqlite3
import pandas as pd
import numpy as np

# 1. DB 연결
db_path = 'chembl_36/chembl_36_sqlite/chembl_36.db' 
conn = sqlite3.connect(db_path)

# 데이터 경로 정의
data_dir = "chembl-data/compound-target"

print("--- Drug-Target Interaction (광범위 추출) ---\n")

query = """
    SELECT
        -- [Drug 정보]
        md.chembl_id AS drug_chembl_id,

        -- [Target 정보]
        td.chembl_id AS target_chembl_id,

        -- [Interaction (Label) 정보]
        act.standard_type AS standard_type,
        act.standard_relation AS standard_relation,
        act.standard_value AS standard_value,
        act.standard_units AS standard_units,
        act.pchembl_value AS pchembl_value,

        -- [메타 데이터]
        ass.confidence_score AS confidence_score,
        ass.description AS assay_description

    FROM
        ACTIVITIES act
    JOIN
        ASSAYS ass ON act.assay_id = ass.assay_id
    JOIN
        TARGET_DICTIONARY td ON ass.tid = td.tid
    JOIN
        MOLECULE_DICTIONARY md ON act.molregno = md.molregno

    WHERE
        md.chembl_id IS NOT NULL
        AND td.chembl_id IS NOT NULL
"""

print("데이터 추출 중...")
df_interactions_all = pd.read_sql_query(query, conn)
conn.close()

# NaN 처리
df_interactions_all = df_interactions_all.replace({np.nan: None})

# 결과 확인
print(f"추출된 데이터 개수: {len(df_interactions_all)}")
print("-" * 60)
print(df_interactions_all.head())
```

```python
import sqlite3
import pandas as pd

db_path = 'chembl_36/chembl_36_sqlite/chembl_36.db' 
conn = sqlite3.connect(db_path)

print("--- [MoA] 임상적으로 검증된 Drug-Target 기전 추출 ---\n")

query = """
    SELECT
        -- [Drug 정보]
        md.chembl_id AS drug_chembl_id,
        md.pref_name AS drug_name,
        md.max_phase AS drug_phase,        -- 약물의 임상 단계

        -- [Target 정보]
        td.chembl_id AS target_chembl_id,
        td.pref_name AS target_name,
        cseq.accession AS uniprot_id,

        -- [MoA 정보] (여기가 핵심!)
        hm.action_type AS action_type,     -- 억제제(INHIBITOR), 작용제(AGONIST) 등
        hm.mechanism_of_action AS description

    FROM
        drug_mechanism hm             -- [테이블 변경] ACTIVITIES 대신 사용
    JOIN
        MOLECULE_DICTIONARY md ON hm.molregno = md.molregno
    JOIN
        TARGET_DICTIONARY td ON hm.tid = td.tid
    JOIN
        TARGET_COMPONENTS tc ON td.tid = tc.tid
    JOIN
        COMPONENT_SEQUENCES cseq ON tc.component_id = cseq.component_id

    WHERE
        td.organism = 'Homo sapiens'
        AND td.target_type = 'SINGLE PROTEIN'
"""

df_moa = pd.read_sql_query(query, conn)
conn.close()

print(f"검증된 기전(MoA) 데이터 개수: {len(df_moa)}")
print("-" * 60)
print(df_moa.head())

# 작용 기전 종류 확인
print("\n[약물 작용 방식 Top 5]")
print(df_moa['action_type'].value_counts().head())
```
