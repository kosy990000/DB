```python
import pandas as pd
```

```python
reactome = pd.read_csv("/data/project/shinyoung/DB/Reactome/reactome-data/NCBI2Reactome_All_Levels.txt", sep="\t", header=None,
                           names=[
        "ensembl_id",
        "reactome_pathway_id",
        "pathway_url",
        "pathway_name",
        "evidence_code",
        "organism"
    ])
```

```python
reactome["organism"].value_counts()
```

```python
reactome_human = reactome[reactome["organism"] == "Homo sapiens"]
reactome_human
```

```python
pathway_table = (
    reactome
    [["reactome_pathway_id", "pathway_name", "organism"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
pathway_ensembl_edge = (
    reactome
    [["reactome_pathway_id", "ensembl_id", "organism", "evidance_code"]]
    .drop_duplicates()
    .reset_index(drop=True)
)
```

```python
pathway_table.to_csv("reactome-data/all_pathway.csv", index=False)
```

```python
pathway_ensembl_edge.to_csv("reactome-data/all_pathway_genes.csv", index=False)
```
