```python
import pandas as pd

# 1. 데이터 로드 (CID를 인덱스로 설정)
print("Loading data...")
# usecols를 사용하여 필요한 컬럼만 추출, 메모리 효율 극대화
df_smiles = pd.read_csv("pubchem-raw-data/CID-SMILES.txt", sep='\t', names=['cid', 'smiles'], index_col='cid')
df_title = pd.read_csv("pubchem-raw-data/CID-Title.txt", sep='\t', names=['cid', 'title'], index_col='cid')
df_inchi = pd.read_csv("pubchem-raw-data/CID-InChI-Key", sep='\t', names=['cid', 'inchi', 'inchi_key'], usecols=['cid', 'inchi_key'], index_col='cid')

# 2. Full Outer Join 실행
print("Performing Full Outer Join...")
# how='outer'를 사용하면 세 파일의 모든 CID 합집합을 구합니다.
combined = df_inchi.join([df_smiles, df_title], how='outer')

# 3. CSV 저장 (빈 값은 빈 칸으로)
print("Saving to CSV...")
combined.to_csv("pudchem-data/pubchem_compound.csv", na_rep='', index=True)

print(f"Total Rows (Union): {len(combined)}")
print("Done!")
```
